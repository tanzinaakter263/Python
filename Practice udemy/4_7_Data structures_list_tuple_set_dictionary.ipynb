**List - Ordered , changeable,allows duplicate and idexable(Mutable)**

```python
# creating a list
fruits=["apple","banana","orange"]
print(fruits)
```

```python
# accessing items using index  (starts at 0)
print(fruits[0]) #apple
print(fruits[1]) #banana
```

```python
#change an element
fruits[2]="mango"
print(fruits)
```

```python
a=[1,2,3.5,False,"word"]
print(a)
```

```python
print(type(a))
```

```python
# add an item to a list
a.append("graps")
print(a)
```

```python
# remove an item from a list
a.remove("word")
print(a)
```

```python
# find len number of elements in a list
print(len(a))
```

```python
# loop through a list
fruits=["apple","banana","orange"]
for fruit in fruits:
  print(fruit)
```

**Tuple - Ordered ,unchangeable,allows duplicate and indexable(Immutable)**

```python
# tuple of colors
colors=("red","green","blue")
print(colors)
print(type(colors))
```

```python
print(colors[0])
print(colors[2])
```

```python
# typing to change an item (this will cause an error)
colors[1]="yellow" #tuples are immutable Error tuple object does not support item assignment
```

```python
for i in colors:
  print(i)
```

```python
a=[1,2,3,4,5]
b=tuple(a)
print(type(a))
print(type(b))
print(a)
print(b)
```

**Set - Unordered ,changeable,not allows duplicate and not indexable and Unique**

```python
# set of student names

students={"Alice","John","Sid"}
print(students)
print(type(students))
```

```python
students.add("Divid")
print(students)
```

```python
students.remove("Sid")
print(students)
```

```python
students.add("Divid1")
print(students)
```

```python
# loop throug set
for i in students:
  print(i)
```

```python
list_1=[1,1,2,3,4,5,5]
print(list_1)
```

```python
set_1={1,1,2,3,4,5,5}   #Remove duplicate values
print(set_1)
```

```python
list_2=[10,9,9,8,5,1,1,2,4,8]
print(list_2)
list_2=set(list_2)
print(list_2)
```

**Dictionary- Key value Data (Mutable) ordered,changeable,not allows duplicate and indexable**

```python
# dictionary with a person details
person={
  "name":"Alice",
  "age":25,
  "is_student":True
}
```

```python
person["age"]
```

```python
person["name"]
```

```python
person["age"]=26
print(person)
```

```python
print(type(person))
```

```python
person["city"]="London"
```

```python
print(person)
```

```python
# loop through a dictionary in for loop
for key,values in person.items():
  print(key,"->",values)
```
