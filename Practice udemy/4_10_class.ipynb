**4.Use class attributes**

Attributes are variables that belog to a class.

```python
# add attributes to a class
class dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
     print("woof!")
```

```python
my_dog=dog(name="Buddy",age=3)
print(my_dog.name)
print(my_dog.age)
my_dog.bark()
```

```python
dog_2=dog(name="Leo",age=4)
print(dog_2.name)
print(dog_2.age)
dog_2.bark()
```

**5.Demostrate inheritance**

Inheritance allows one class to use the properties and methods of another class.

```python
# add attributes to a class
class dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
     print("woof!")
```

```python
# create a sunclass that inherites from dog
class Puppy(dog):
  def play(self):
    print(self.name,"is playful")
```

```python
my_puppy=Puppy(name="Max",age=1)
```

```python
print(my_puppy.name)
print(my_puppy.age)
my_puppy.bark()
my_puppy.play()
```
