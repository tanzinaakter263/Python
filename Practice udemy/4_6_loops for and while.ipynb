**2.While loop**

```python
count=1
while count<=5:
  print("Count is: ",count)
  count+=1
```

```python
# Be careful this loop never stops - stop with the stop symbol on the cell
while True:
  print("This is an infinite loop")
```

**3.Loop controls - break and continue**

```python
# break -  stop a loop when condition is reached
for i in range(0,10):
  if i==3:
    break
  print(i)
```

```python
# continue - skip an iteration
for i in range(0,10):
  if i==3:
    continue
  print(i)
```
