```python
x=10
print(x)
print(type(x))

# float(var) -> convert to float
x=float(x)
print(x)
print(type(x))
```

```python
y=10.5
print(y)
print(type(y))

# int(var) -> convert to integer
y=int(y)
print(y)
print(type(y))
```
